```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.image import imread
```

```python
# data 준비
x = np.arange(0, 6, 0.1) # 0~6까지 0.1 간격으로 생성
y = np.sin(x)
```

```python
#그래프 그리기
plt.plot(x, y)
plt.show()
```

```python
x = np.arange(0, 6, 0.1)

y1 = np.sin(x)
y2 = np.cos(x)

plt.plot(x, y1, label = "sin")
plt.plot(x, y2, linestyle = "--", label = "cos")
plt.xlabel("x")
plt.ylabel("y")
plt.title("sin & cos")
plt.legend()
plt.show()
```
