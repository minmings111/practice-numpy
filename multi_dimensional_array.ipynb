```python
X = np.array([1.0, 0.5])
W1 = np.array([[0.1, 0.3, 0.5], [0.2, 0.4, 0.6]])
B1 = np.array([0.1, 0.2, 0.3])

print(X.shape)
print(W1.shape)

# 1차원 배열은 가로, 세로라는 방향의 개념 없이 1차원 구조라 (3,) 구조로 출력됨.
print(B1.shape)

A1 = np.dot(X, W1) + B1
A1
```

```python
# 활성화 함수: 시그모이드
# 입력층(0층) -> 은닉층(1층)
def sigmoid(x):
  return 1 / (1 + np.exp(-x))


Z1 = sigmoid(A1) # 1층의 최종 결과값

print(A1)
print(Z1)
```

```python
# 1층 -> 2층
W2 = np.array([[0.1, 0.4], [0.2, 0.5] , [0.3, 0.6]])
B2 = np.array([0.1, 0.2])

print(W2.shape)
print(B2.shape)

print(Z1.shape)

A2 = np.dot(Z1, W2) + B2 # 1층의 최종 결과값이 인자가 됨
Z2 = sigmoid(A2) # 2층의 최종 결과값

print(A2)
print(Z2)
```

```python
# 출력층

# 출력층의 활성화 함수는 풀고자 하는 문제의 성질에 맞추어 선택함.
# 활성화 함수: 항등 함수(입력을 그대로 출력하는 함수)
def identity_function(x):
  return x

W3 = np.array([[0.1, 0.3], [0.2, 0.4]])
B3 = np.array([0.1, 0.2])

A3 = np.dot(Z2, W3) + B3

Y = identity_function(A3)

print(Y)
```

```python
# 구현 정리
def sigmoid(x):
  return 1 / (1 + np.exp(-x))

def identity_function(x):
  return x

def init_network():
  network = {} # 빈 딕셔너리 선언

  network["W1"] = np.array([[0.1, 0.3, 0.5], [0.2, 0.4, 0.6]])
  network["b1"] = np.array([0.1, 0.2, 0.3])
  network["W2"] = np.array([[0.1, 0.4], [0.2, 0.5] , [0.3, 0.6]])
  network["b2"] = np.array([0.1, 0.2])
  network["W3"] = np.array([[0.1, 0.3], [0.2, 0.4]])
  network["b3"] = np.array([0.1, 0.2])

  return network

def forward(network, x):
  W1, W2, W3 = network["W1"], network["W2"], network["W3"]
  b1, b2, b3 = network["b1"], network["b2"], network["b3"]

  a1 = np.dot(x, W1) + b1
  z1 = sigmoid(a1)
  a2 = np.dot(z1, W2) + b2
  z2 = sigmoid(a2)
  a3 = np.dot(z2, W3) + b3
  y = identity_function(a3)

  return y

network = init_network() # 네트워크 변수 초기화
x = np.array([1.0, 0.5]) # 최초 x 인자 전달
y = forward(network, x) # 순방향 계산
print(y)
```
